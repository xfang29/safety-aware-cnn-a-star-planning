```python
import sys
from pathlib import Path

import numpy as np
```

```python
def find_project_root() -> Path:
    """Find the project directory containing the src folder."""
    current_directory = Path.cwd().resolve()

    for candidate in [
        current_directory,
        current_directory.parent,
    ]:
        if (candidate / "src").is_dir():
            return candidate

    raise RuntimeError(
        "Could not find the project root containing src/."
    )


PROJECT_ROOT = find_project_root()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(
        0,
        str(PROJECT_ROOT),
    )

print("Project root:", PROJECT_ROOT)
```

```python
from src.dataset_generation import (
    generate_dataset_split,
    load_dataset_split,
    save_dataset_split,
)
```

```python
smoke_split = generate_dataset_split(
    seeds=range(10),
    label_sigma=2.0,
    progress_every=5,
)

print("Inputs:", smoke_split.inputs.shape)
print("Soft labels:", smoke_split.soft_labels.shape)
print("Path masks:", smoke_split.path_masks.shape)
print("Starts:", smoke_split.starts.shape)
print("Goals:", smoke_split.goals.shape)
print("Seeds:", smoke_split.seeds.shape)
print("Expert lengths:", smoke_split.expert_lengths.shape)
```

```python
smoke_path = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "smoke_test.npz"
)

saved_path = save_dataset_split(
    smoke_split,
    smoke_path,
)

loaded_smoke_split = load_dataset_split(
    saved_path
)

print("Saved to:", saved_path)
print("Loaded size:", loaded_smoke_split.size)

print(
    "Inputs identical:",
    np.array_equal(
        smoke_split.inputs,
        loaded_smoke_split.inputs,
    ),
)

print(
    "Soft labels identical:",
    np.array_equal(
        smoke_split.soft_labels,
        loaded_smoke_split.soft_labels,
    ),
)

print(
    "Path masks identical:",
    np.array_equal(
        smoke_split.path_masks,
        loaded_smoke_split.path_masks,
    ),
)

print(
    "Seeds identical:",
    np.array_equal(
        smoke_split.seeds,
        loaded_smoke_split.seeds,
    ),
)
```

```python
train_seeds = range(0, 800)
validation_seeds = range(10_000, 10_100)
test_seeds = range(20_000, 20_100)
```

```python
train_split = generate_dataset_split(
    seeds=train_seeds,
    label_sigma=2.0,
    progress_every=100,
)

train_path = save_dataset_split(
    train_split,
    PROJECT_ROOT
    / "data"
    / "processed"
    / "train.npz",
)

print("Train set saved to:", train_path)
print("Train inputs:", train_split.inputs.shape)
print("Train labels:", train_split.soft_labels.shape)
```

```python
validation_split = generate_dataset_split(
    seeds=validation_seeds,
    label_sigma=2.0,
    progress_every=50,
)

validation_path = save_dataset_split(
    validation_split,
    PROJECT_ROOT
    / "data"
    / "processed"
    / "validation.npz",
)

print(
    "Validation set saved to:",
    validation_path,
)
print(
    "Validation inputs:",
    validation_split.inputs.shape,
)
```

```python
test_split = generate_dataset_split(
    seeds=test_seeds,
    label_sigma=2.0,
    progress_every=50,
)

test_path = save_dataset_split(
    test_split,
    PROJECT_ROOT
    / "data"
    / "processed"
    / "test.npz",
)

print("Test set saved to:", test_path)
print("Test inputs:", test_split.inputs.shape)
```
